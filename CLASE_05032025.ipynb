```python
df_hurto_personas= pd.read_csv("/content/drive/MyDrive/Analisis de datos/Booth_Camp/Datasets/hurto_a_persona.csv",low_memory=False)
```

```python
##histogramas de hurto personas por edades
plt.figure(figsize=(5,2))
plt.hist(df_hurto_personas["edad"])
```

```python
##creacion histograma para medios de transporte
plt.figure(figsize=(5,2))
plt.hist(df_hurto_personas["medio_transporte"]) ##para la orientacion de los ejes se añade la propiedad dentro de la orden de la funcion
plt.xticks(rotation=90) #rotar etiqueta del eje x
```

```python
# Crear histograma por medio de transporte
sns.histplot(data=df_hurto_personas,
             x="medio_transporte" , #indico eje
             hue="sexo", #categorias que visualizo
             multiple="stack") ## apilar categorias
plt.xticks(rotation=75)
```

```python
#histograma modalidad de hurto
plt.hist(df_hurto_personas["modalidad"])
plt.xticks(rotation=90)
plt.show()
```

```python
sns.displot(df_hurto_personas,
            x="edad",
            hue="medio_transporte",
            multiple="stack"
            )
```

```python
sns.displot(df_hurto_personas,
            x="edad",
            hue="modalidad",
            multiple="stack"
            )
```

```python
plt.savefig("coeff.svg") ##guardar en un formato de alta definición
```

```python
##caja y bigotes para relaciones entre sexo y edad
plt.figure(figsize=(8,5))
sns.boxplot(x=df_hurto_personas["sexo"], y=df_hurto_personas["edad"], palette="pastel") ##x tomo campo categorico ## y tomo una variable numerica ## configuro estilo de grafico
plt.title("Distribucion en hurto de personas para edades por sexo", fontsize=14)
plt.xlabel("Sexo", fontsize=12)
plt.ylabel("Edad", fontsize=12)
plt.xticks(rotation=45)
plt.show()
plt.savefig("Dist_edad_sexo.svg")
```

```python
#Diagrama de cajas de estado civil segun edad
plt.figure(figsize=(8,5))
sns.boxplot(
    x=df_hurto_personas["estado_civil"],
    y=df_hurto_personas["edad"],
    palette="pastel")
plt.title(
    "Distribución en hurto de personas para estado civil por edades",
    fontsize=14)
plt.xlabel("Estado civil", fontsize=12)
plt.ylabel("Edad", fontsize=12)
plt.xticks(rotation=90)
plt.show()
```

```python
sns.catplot(
    data=df_hurto_personas,
    x="estado_civil",
    y="edad",
    hue="sexo",
    kind="bar") ## Establesco si deseo apilar una categoria sobre otra
plt.xticks(rotation=90)
plt.show()
```

```python
##grafica por modalidad de hurto en categorias de  edad y  haciendo enfasis en sexo
plt.figure(figsize=(40,30))
sns.catplot(
    data=df_hurto_personas,
    x="modalidad",
    y="edad",
    hue="sexo",
    kind="bar",
    aspect=4,
    width=0.5,
    dodge=True
)
plt.xticks(rotation=90)
plt.savefig("/content/drive/MyDrive/Analisis de datos/Booth_Camp/graficos/prueba.png",dpi=300, bbox_inches="tight")
plt.show()
```

```python
##grafica por modalidad de hurto en categorias de  edad y  haciendo enfasis en sexo
plt.figure(figsize=(40,10))
sns.catplot(
    data=df_hurto_personas.head(10),
    x="modalidad",
    y="edad",
    hue="sexo",
    kind="bar"
)
plt.xticks(rotation=90)
plt.show()
```
